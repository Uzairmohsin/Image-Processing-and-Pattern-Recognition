```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import ndimage
```

```python
from matplotlib.image import imread

image = imread(r"C:\Users\uzair\Downloads\My_photo.jpeg")
    

if len(image.shape) == 3:
    image = image.mean(axis=2)

plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis('off')
```

convert to binary

```python
binary = image > 127

plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis('off')
```

define structuring element

```python
structure = np.ones((5,5))
```

erosion

```python
erosion = ndimage.binary_erosion(binary, structure=structure)

plt.imshow(erosion, cmap='gray')
plt.title("Erosion")
plt.axis('off')
```

dialation

```python
dilation = ndimage.binary_dilation(binary, structure=structure)

plt.imshow(dilation, cmap='gray')
plt.title("Dilation")
plt.axis('off')
```

opening    

```python
opening = ndimage.binary_opening(binary, structure=structure)

plt.imshow(opening, cmap='gray')
plt.title("Opening")
plt.axis('off')
```

closing

```python
closing = ndimage.binary_closing(binary, structure=structure)

plt.imshow(closing, cmap='gray')
plt.title("Closing")
plt.axis('off')
```

all results

```python
titles = ['Original', 'Binary', 'Erosion', 'Dilation', 'Opening', 'Closing']
images = [image, binary, erosion, dilation, opening, closing]

plt.figure(figsize=(10,6))
for i in range(6):
    plt.subplot(2,3,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

boundry detection

```python
structure = np.ones((3,3))
```

erosion and dialation

```python
erosion = ndimage.binary_erosion(binary, structure=structure)
dilation = ndimage.binary_dilation(binary, structure=structure)
```

inner boundry

```python
inner_boundary = binary.astype(int) - erosion.astype(int)
```

outer boundry

```python
outer_boundary = dilation.astype(int) - binary.astype(int)
```

```python
titles = ["Original", "Inner Boundary", "Outer Boundary"]
images = [binary, inner_boundary, outer_boundary]

plt.figure(figsize=(10,4))

for i in range(3):
    plt.subplot(1,3,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')     

plt.tight_layout()
plt.show()
```
